#### Tokenize the Corpus using the nltk
Tokenization is the process of breaking down text into smaller units called "tokens." These tokens can be words, subwords, characters, or even specific patterns like punctuation marks.

```python
# Tokenize the Corpus using the nltk.word_tokenize()
bbc_tokens = nltk.word_tokenize(bbc_corpus)
# The output of nltk.word_tokenize() is each individual word or punctuation mark in the text, treated as separate tokens. 
# It does not split into single characters, but rather into meaningful components like words, numbers, and symbols.
```

```python
# 1)
# Calculate the number of tokens in the corpus. 
# Think about it as length of a list.
print("The number of tokens in the corpus is:", len(bbc_tokens))
```

```python
# 2)
# Calculate the number of unique tokens (types) there are in the corpus. 
# "This is a token. This is another token." There are two occurancies of "token" in the previous
# string but one type "token".
# To calculate the types think about the set() function
print("The number of unique tokens is:", len(set(bbc_tokens)))
```

#### Stopwords
In Natural Language Processing (NLP), stopwords are commonly used words in a language that typically don't add significant meaning to the analysis of text.  
Examples in English include words like "the," "is," "in," "and," etc. These words often appear frequently in text but are usually ignored during processing because they don't contribute much to understanding the core semantics of the text.

```python
# Import the predefined stopwords in NLTK
from nltk.corpus import stopwords
stop_words = stopwords.words('english') # The list of English stopwords!
```

```python
# 3)
# How many stopword are there in the stop word list? 
# Are this few or many stop words, can you think about other stopwords?
print("The number of stopwords in the predefined stopwords is:", len(stop_words))
```

```python
stopword_count_in_BBC = sum(1 for word in bbc_tokens if word.lower() in stop_words)
print(f"Number of stopwords in the BBC corpus: {stopword_count_in_BBC}")
```

```python
# Write a Python function that filters out the stopwords from a corpus
def get_corpus_without_stopwords(tokens):
    clean_corpus = []
    for word in tokens:
        if(word.lower() not in stop_words):
            clean_corpus.append(word)
    return clean_corpus
```

```python
bbc_clean_corpus = get_corpus_without_stopwords(bbc_tokens)
```

```python
# How many tokens are there in the BBC corpus after filter out the stopwords
# are these reasonable or to many or too few?
print("The number of tokens in the BBC corpus is:", len(bbc_tokens))
print(f"Number of stopwords in the BBC corpus is: {stopword_count_in_BBC}")
print("The number of tokens in BBC_corpus after filtering out ALL the stopwords:", (len(bbc_corpus) - len(bbc_clean_corpus)))
```

#### Lemmatizer
A lemmatizer is a tool in natural language processing (NLP) that reduces words to their base or canonical form, known as the lemma. This process is called lemmatization, and it helps transform different forms of a word (like plural, past tense, etc.) into a single standardized representation.  

Example: The words "running," "ran," and "runs" can all be reduced to their lemma, "run."

```python
# Import the WordNet lemmatizer in NLTK
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
# Write a Python function that lemmatizes the tokens in the BBC corpus
def lemmatize_corpus(tokens):
    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in tokens]
    return ' '.join(lemmatized_tokens)
```

```python
bbc_lemmatized = lemmatize_corpus(bbc_clean_corpus)
print(bbc_lemmatized[:200])
```

```python
# 7. How many types do you have in the BBC corpus after lemmatization?
print("The number of unique tokens:", len(set(bbc_lemmatized)))
```

#### Stemmer
A stemmer is a tool in natural language processing (NLP) that reduces a word to its root or stem form. Unlike a lemmatizer, which relies on linguistic rules to derive the base form (the lemma), a stemmer operates using heuristics and typically trims off common prefixes or suffixes. As a result, the stem may not always be a valid word, but it’s a simplified version used to group similar word forms together.  

Example: The words "studies" and "studying" might become "studi."

```python
stemmer = nltk.stem.PorterStemmer()
```

```python
# Write a Python function that stems the tokens in the BBC corpus
def stem_corpus(tokens):
    stemmed_tokens = [stemmer.stem(token) for token in tokens]
    return ' '.join(stemmed_tokens)
```

```python
# 8)
# How many types do you have in the BBC Corpus after stemming?
# Are there any differences? If so, why?
bbc_stemmed = stem_corpus(bbc_clean_corpus)
print(bbc_stemmed[:200])
```

```python
print("The number of unique tokens:", len(set(bbc_stemmed)))
# The difference is that stemming is a more aggressive process than lemmatization.
```

### Now we are ready with the corpus linguistic part of the laboration exercise and we start with POS taggning and Chunking which is more NLP processing tasks.

```python
# Use the nltk.pos_tag() function to POS tag the BBC corpus
from nltk import pos_tag
from nltk import ne_chunk
```

```python
# 9) Inspect the tagged corpus, what types of tags do you see? What do they mean?
tagged_corpus = pos_tag(bbc_clean_corpus)
```

```python
# Chunking
chunked_corpus = ne_chunk(tagged_corpus)
```

```python
#Print some named entities in the sentence
for subtree in chunked_corpus:
    if hasattr(subtree, 'label'):
        print(subtree.label(), ' '.join(c[0] for c in subtree.leaves()))
```

```python
#Print some named entities in the sentence, #PERSON, ORGANIZATION and GPE
ctr = 0
for subtree in chunked_corpus:
    if hasattr(subtree, 'label'):
        if subtree.label() in ['PERSON', 'ORGANIZATION', 'GPE']:
            print(subtree.label(), ' '.join(c[0] for c in subtree.leaves()))
            ctr += 1
            if ctr > 10:
                break
```
